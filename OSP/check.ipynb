```python
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------
# LOAD YOUR CSV / XLSX DATA
# -------------------------
# Example if file is CSV:
# df = pd.read_csv("your_file.csv")

# Example if file is Excel:
df = pd.read_excel("osteoporosis_predictions.xlsx")



df = pd.DataFrame(df)
df["Confidence"] = pd.to_numeric(df["Confidence"], errors="coerce")

# -------------------------
# HISTOGRAM PER TRUE CLASS
# -------------------------
plt.figure(figsize=(10,5))

for cls in df["Actual Class"].unique():
    subset = df[df["Actual Class"] == cls]["Confidence"]
    plt.hist(subset, bins=30, alpha=0.6, label=f"{cls} (n={len(subset)})")

plt.xlabel("Model Confidence")
plt.ylabel("Frequency")
plt.title("Confidence Distribution per True Class")
plt.legend()
plt.tight_layout()
plt.show()


# -------------------------
# SCATTERPLOT (INDEX vs CONFIDENCE)
# -------------------------
plt.figure(figsize=(10,5))

for cls in df["Actual Class"].unique():
    subset = df[df["Actual Class"] == cls]
    plt.scatter(subset.index, subset["Confidence"], label=cls)

plt.xlabel("Sample Index")
plt.ylabel("Confidence")
plt.title("Scatterplot: Confidence per Sample (Colored by True Class)")
plt.legend()
plt.tight_layout()
plt.show()
```
